```python
import numpy as np
import matplotlib.pyplot as plt

# Dataset AND
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([0, 0, 0, 1])

# Parametri iniziali
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 10

# Funzione di attivazione
def activation(x):
    return 1 if x >= 0 else 0

# Addestramento
for epoch in range(epochs):
    for i in range(len(X)):
        input_vec = X[i]
        target = y[i]
        linear_output = np.dot(input_vec, weights) + bias
        prediction = activation(linear_output)
        error = target - prediction
        weights += learning_rate * error * input_vec
        bias += learning_rate * error

# Visualizzazione
plt.figure(figsize=(6, 6))

# Punti
for i in range(len(X)):
    color = 'red' if y[i] == 0 else 'blue'
    plt.scatter(X[i][0], X[i][1], color=color, s=100, edgecolors='k')

# Retta di decisione: w1*x + w2*y + b = 0 → y = -(w1*x + b)/w2
x_vals = np.linspace(-0.2, 1.2, 100)
if weights[1] != 0:
    y_vals = -(weights[0] * x_vals + bias) / weights[1]
    plt.plot(x_vals, y_vals, 'g--', label='Decision boundary')
else:
    plt.axvline(-bias / weights[0], color='g', linestyle='--', label='Decision boundary')

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Percettrone – Funzione AND')
plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)
plt.grid(True)
plt.legend()
plt.show()
```
