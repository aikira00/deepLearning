```python
# Reuters Dataset - Esempio didattico
from tensorflow.keras.datasets import reuters
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

# Caricamento del dataset (limitiamo a top 10.000 parole più frequenti)
num_words = 10000
(x_train, y_train), (x_test, y_test) = reuters.load_data(num_words=num_words)

print(f"Numero di campioni di training: {len(x_train)}")
print(f"Numero di classi target: {np.max(y_train) + 1}")  # +1 perché parte da 0

# Ogni input è una sequenza di indici interi
print("\nEsempio di sequenza di input (primo campione):")
print(x_train[0])

# ---- Dizionario Parole → Indici ----
word_index = reuters.get_word_index()

# Invertiamo il dizionario per passare da indice a parola
reverse_word_index = {value + 3: key for key, value in word_index.items()}

# Nota: i primi 3 indici (0, 1, 2) sono riservati per:
# 0 = padding, 1 = start of sequence, 2 = unknown word

reverse_word_index[0] = '[PAD]'
reverse_word_index[1] = '[START]'
reverse_word_index[2] = '[UNK]'

# ---- Funzione per decodificare una sequenza di indici in una frase ----
def decode_review(sequence):
    return ' '.join([reverse_word_index.get(i, '?') for i in sequence])

# ---- Esempio: Mostriamo il testo della prima notizia ----
print("\nTesto della prima notizia (decodificato):")
print(decode_review(x_train[0]))

# ---- Se vuoi vedere una parola dato un indice ----
indice = 10
print(f"\nParola associata all'indice {indice}:")
print(reverse_word_index.get(indice, 'Indice non presente'))

# ---- Se vuoi vedere l'indice dato una parola ----
parola = 'oil'
print(f"\nIndice associato alla parola '{parola}':")
print(word_index.get(parola, 'Parola non presente'))
```

```python
from gensim.models import Word2Vec
from gensim.test.utils import common_texts

# Addestriamo un piccolo modello Word2Vec sui dati di esempio
model = Word2Vec(sentences=common_texts, vector_size=10, window=5, min_count=1, workers=1)

# ---- Recupero embedding di una parola presente ----
word = 'computer'
if word in model.wv:
    embedding_vector = model.wv[word]
    print(f"\nEmbedding per la parola '{word}':\n{embedding_vector}")
else:
    print(f"\nLa parola '{word}' non è nel vocabolario del modello.\n")

# ---- Parole più simili a una parola data ----
word_query = 'human'
print(f"\nParole più simili a '{word_query}':")
for similar_word, similarity in model.wv.most_similar(word_query):
    print(f"{similar_word} → similarità: {similarity:.4f}")

# ---- Similarità tra due parole ----
word1 = 'human'
word2 = 'computer'
similarity = model.wv.similarity(word1, word2)
print(f"\nSimilarità tra '{word1}' e '{word2}': {similarity:.4f}")
```

```python
from gensim.models import KeyedVectors

# Caricamento del modello Word2Vec pre-addestrato
path = 'GoogleNews-vectors-negative300.bin.gz'
model = KeyedVectors.load_word2vec_format(path, binary=True)

# Vediamo la similarità tra 'apple' e alcune parole
similar_fruit = model.similarity('apple', 'fruit')
similar_iphone = model.similarity('apple', 'iphone')
similar_google = model.similarity('apple', 'google')

print(f"Similarità tra 'apple' e 'fruit': {similar_fruit:.4f}")
print(f"Similarità tra 'apple' e 'iphone': {similar_iphone:.4f}")
print(f"Similarità tra 'apple' e 'google': {similar_google:.4f}")

# Troviamo le parole più simili ad 'apple'
print("\nParole più simili ad 'apple':")
for word, score in model.most_similar('apple', topn=10):
    print(f"{word}: {score:.4f}")
```
