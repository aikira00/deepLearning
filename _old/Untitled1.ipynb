```python
print("=== ESERCIZI INTERATTIVI: SCOPRI COSA SUCCEDE IN PYTHON! ===\n")

# Esercizio 1: Apici mancanti
print("\n#1 - Cosa succede se dimentichi un apice in una stringa?")
print('print("Ciao mondo)')  # <-- NON ESEGUIRE! Copia e prova tu a mano
input("Scrivi cosa succede quando esegui questo codice: ")

# Esercizio 2: Parentesi mancanti
print("\n#2 - Cosa succede se dimentichi una parentesi in print?")
print('print("Benvenuto!"')  # <-- Copia e prova a eseguire
input("Scrivi cosa succede: ")

# Esercizio 3: Doppi segni più
print("\n#3 - Cosa succede con +2 o 2++2?")
print("Esempio: print(+2) ➜ Risultato:", +2)
print("Esempio: print(2++2) ➜ Risultato:", 2++2)
input("Perché funziona? Prova a spiegarlo: ")

# Esercizio 4: Zeri iniziali
print("\n#4 - Posso scrivere numero = 042?")
print("Prova a scrivere numero = 042 in un tuo file")
input("Cosa dice Python quando esegui il codice? ")

# Esercizio 5: Due numeri senza operatore
print("\n#5 - Cosa succede se scrivi: print(4 5)?")
print("Copia il codice in un file e prova:")
print("    print(4 5)")
input("Scrivi che errore ti dà: ")

# Esercizio 6: Assegnamento al contrario
print("\n#6 - Cosa succede se scrivi: 42 = n?")
print("Prova in un file Python:")
print("    42 = n")
input("Python lo accetta? Perché no? ")

# Esercizio 7: x = y = 1
print("\n#7 - Cosa succede con: x = y = 1?")
x = y = 1
print("x =", x, ", y =", y)
input("Entrambe le variabili hanno ricevuto 1? ")

# Esercizio 8: Punto e virgola
print("\n#8 - Posso mettere il punto e virgola?")
print("Eseguo:")
print("    print('Ciao'); print('Tutto bene?')")
print('Risultato:')
print("Ciao"); print("Tutto bene?")
input("Python lo accetta? Serve a qualcosa? ")

# Esercizio 9: Parole vietate
print("\n#9 - Posso usare 'class' o 'print' come nome di variabile?")
print("Prova questo codice:")
print("    class = 5")
print("    print = 3")
input("Cosa ti segnala Python? ")

# Esercizio 10: Niente indentazione
print("\n#10 - E se dimentico l’indentazione in un blocco?")
print("Scrivi questo codice:")
print('''
x = 10
if x > 5:
print("Maggiore di 5")
''')
input("Quale errore ricevi? ")

# Esercizio 11: Variabili con spazi
print("\n#11 - Posso usare uno spazio in una variabile?")
print("Prova a scrivere:")
print('    nome completo = "Mario"')
input("Errore? Quale? ")

# Esercizio 12: Virgola nei numeri
print("\n#12 - Posso scrivere: prezzo = 9,99?")
prezzo = 9,99
print("Eseguendo: prezzo = 9,99 ➜ prezzo =", prezzo)
input("È quello che ti aspettavi? ")

print("\n== Ottimo lavoro! Continua a sperimentare! ==")
```
